## Insights from Average Temperature Analysis

- Coastal cities such as Mumbai, Thane, Navi Mumbai, Vasai Virar, and Bhiwandi recorded the highest average temperatures (~28.34°C). This may be influenced by humid coastal climate conditions.

- Satara (24.37°C) and Nashik (24.79°C) showed the lowest average temperatures among the cities in the dataset, indicating comparatively cooler climates.

- Most Maharashtra cities have average temperatures between 25°C and 27°C, showing relatively stable temperature patterns across the state.

- Akola (27.50°C), Chandrapur (27.42°C), and Solapur (27.27°C) are among the hottest inland cities in the dataset.

- Pune and Pimpri Chinchwad share the same average temperature (25.45°C), which is expected because both cities are geographically close.

- Several nearby cities such as Jalgaon and Bhusawal also show nearly identical temperature averages, suggesting similar regional climate conditions.

- The dataset suggests that coastal regions generally experience higher average temperatures than hilly or elevated regions like Satara and Nashik.

## Insights from Rainfall (Precipitation) Analysis

- Mumbai, Thane, Navi Mumbai, Vasai Virar, and Bhiwandi recorded the highest total precipitation (~17,883 mm), indicating extremely high rainfall in coastal regions of Maharashtra.

- Nashik (4452.3 mm), Dhule (4459.1 mm), and Malegaon (4459.1 mm) recorded the lowest precipitation values in the dataset.

- Coastal cities generally receive much higher rainfall compared to inland cities due to the influence of the Arabian Sea and monsoon winds.

- Nanded (8919.4 mm), Nagpur (8503.6 mm), Wardha (8205.9 mm), and Chandrapur (8066.1 mm) showed relatively high rainfall among inland cities.

- Pune and Pimpri Chinchwad recorded the same precipitation values (6205 mm), which is expected because both cities are geographically connected.

- The analysis reveals a strong regional rainfall pattern where western coastal cities receive significantly more precipitation than northern and central regions.

```python
import matplotlib.pyplot as plt
df.groupby("year")["tavg"].mean().plot(kind="bar", color="lightblue")
plt.title("Yearly Average Temperature")
plt.ylabel("Average Temperature")
plt.xlabel("Year")
plt.ylim(24, 28)
plt.tight_layout()
plt.show()
```

Maharashtra's average temperature remained consistently between 26-27°C from 2020-2025, showing climate stability across the region with a slight dip in 2025.

```python
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
ax = df.groupby("month")["tavg"].mean().plot(kind="bar", color="lightpink")
ax.set_xticklabels(month_names, rotation=45)
plt.title("Monthly Average Temperature in Maharashtra")
plt.ylabel("Average Temperature (°C)")
plt.xlabel("Month")
plt.ylim(20, 35)
plt.tight_layout()
plt.show()
```

Maharashtra experiences peak temperatures in April and May (30-31°C), followed by a cooling effect during monsoon months (July-September) dropping to 26°C, and coldest temperatures in January and December (23°C) — showing a clear seasonal temperature cycle.

```python
df.groupby("city")["tavg"].mean().sort_values(ascending=False)
```

```python
df.groupby("city")["tavg"].mean().sort_values(ascending=False).head(10).plot(kind="bar", color="coral")
plt.title("Top 10 Hottest Cities in Maharashtra")
plt.ylabel("Average Temperature (°C)")
plt.xlabel("City")
plt.ylim(24, 30)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Coastal cities of Maharashtra — Navi Mumbai, Thane, Mumbai, Bhiwandi and Vasai Virar — dominate the top 5 hottest cities by average temperature (28.5°C), suggesting that coastal proximity maintains consistently higher temperatures throughout the year compared to inland cities like Akola, Chandrapur and Osmanabad which appear in the lower half of the top 10.

```python
df.groupby("city")["tmin"].mean().sort_values().head(10).plot(kind="bar", color="lightblue")
plt.title("Top 10 Coldest Cities in Maharashtra")
plt.ylabel("Min Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Pune and Pimpri Chinchwad record the lowest minimum temperatures in Maharashtra (around 18.7°C), followed by Satara and Nashik — all of which are inland plateau cities at higher elevations. This contrasts sharply with coastal cities, confirming that elevation and distance from the coast are key factors in determining minimum temperatures across Maharashtra.

```python
df.groupby("city")["tmax"].max().sort_values(ascending=False)
```

Inland cities like Achalpur, Akola and Amravati record the highest peak temperatures in Maharashtra (44°C), while coastal cities like Mumbai, Thane and Navi Mumbai never exceed 35°C — confirming that the Arabian Sea keeps coastal areas significantly cooler during extreme heat events despite having higher year-round averages.

```python
df.groupby("month")["pres"].mean().plot(kind="line", marker="o", color="purple")
plt.title("Monthly Average Atmospheric Pressure in Maharashtra")
plt.xlabel("Month")
plt.ylabel("Pressure (hPa)")
plt.xticks(range(1, 13), ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
plt.tight_layout()
plt.show()
```

Atmospheric pressure in Maharashtra follows a perfect V-shaped seasonal pattern — starting high in January (1013 hPa), steadily dropping to its lowest point in July (1004 hPa) which coincides exactly with peak monsoon rainfall, then recovering back to winter highs by December. This inverse relationship between pressure and rainfall confirms that low pressure systems are the primary driver of Maharashtra's monsoon season.

```python
monthly_rain = df.groupby("month")["prcp"].mean()

month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

monthly_rain.plot(kind="line", marker="o", color="blue")
plt.title("Average Monthly Rainfall in Maharashtra")
plt.xlabel("Month")
plt.ylabel("Rainfall (mm)")
plt.xticks(range(1, 13), month_names)
plt.tight_layout()
plt.show()
```

Maharashtra receives majority of its rainfall during monsoon months June and July, with July peaking at nearly 480mm — while remaining months from October to May receive almost negligible rainfall, confirming Maharashtra's strong dependence on a 3-4 month monsoon window for its annual water supply.

```python
yearly_temp = df.groupby("year")["tavg"].mean()

yearly_temp.plot(kind="line", marker="o")
plt.title("Yearly Temperature Trend")
plt.xlabel("Year")
plt.ylabel("Average Temperature (°C)")
plt.show()
```

Maharashtra's average temperature showed a declining trend from 27°C in 2020 to 26.5°C in 2022, followed by a recovery peaking at 26.95°C in 2024 — suggesting a cyclical temperature pattern across the state over the past 5 years.

```python
df["tavg"].hist(color="steelblue", edgecolor="black")
plt.title("Temperature Distribution in Maharashtra")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Frequency (Number of Records)")
plt.tight_layout()
plt.show()
```

Maharashtra's average temperature follows a normal distribution, with most readings concentrated between 25-28°C — indicating that moderate warm temperatures are the most common climate experience across the state, with extreme cold below 20°C and extreme heat above 33°C being relatively rare.

## Final Conclusion:
After analyzing Maharashtra's weather data across 36 cities from 2020 to 2025, several clear and interesting patterns emerged across temperature, rainfall and atmospheric pressure.
Temperature analysis revealed a strong contrast between coastal and inland cities. Coastal cities like Mumbai, Navi Mumbai and Thane maintain consistently higher average temperatures throughout the year due to the moderating influence of the Arabian Sea. However inland cities like Achalpur, Akola and Amravati recorded extreme peak temperatures reaching up to 44°C — nearly 9°C higher than any coastal city's maximum. On the other end, plateau cities like Pune, Pimpri Chinchwad and Satara recorded the lowest minimum temperatures around 18.7°C, showing that elevation plays a significant role in keeping these cities cooler.
Maharashtra's temperature follows a clear seasonal cycle — rising sharply from March, peaking in April-May around 30-31°C, cooling during monsoon months and recovering post-monsoon. The overall temperature distribution shows 25-28°C is the most frequently experienced range across the state.
Rainfall analysis confirmed Maharashtra's complete dependence on monsoon — July alone averages nearly 480mm while most other months receive close to zero rainfall, making the 3-4 month monsoon window critical for the entire state's water needs.
Most interestingly, atmospheric pressure showed a perfect V-shaped seasonal pattern — dropping from 1013 hPa in January to its lowest point of 1004 hPa in July, directly coinciding with peak monsoon rainfall. This confirms that low pressure systems are the primary driver of Maharashtra's monsoon season.
Overall this analysis paints a comprehensive picture of Maharashtra's diverse climate — shaped by geography, elevation, coastal proximity and seasonal pressure systems working together.
